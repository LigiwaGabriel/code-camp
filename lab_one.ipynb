```python
import pandas as pd
```

## Pandas Series 
- A series is a 1D array. This means it contains a single column 

```python
## creating a series data.
student_scores = pd.Series(
    [89,98,79,79,69,56],
    name="student scores",
    index=["Jane","John","James","Joseph","Josephine","Andrew"]
)
student_scores
```

```python
student_data={
    "Name":["Jane","John","James","Joseph","Josephine","Andrew"],
    "Science":[89,98,79,79,69,56],
    "Maths":[91,86,95,69,89,66],
    "Gender":["Female","Male","Male","Male","Female","Male"]
}
students=pd.DataFrame(student_data)
students
```
