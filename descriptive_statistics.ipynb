## MEASURES OF CENTRAL TENDENCY
They only three for example: 
- Mean
- Median 
- Mode 


```python
import pandas as pd 

student_data={
    "Name":["Jane","John","James","Joseph","Josephine","Andrew"],
    "Science":[89,98,79,79,69,56],
    "Maths":[91,86,95,69,89,66],
    "Gender":["Female","Male","Male","Male","Female","Male"]
}
students=pd.DataFrame(student_data)
students
```

```python
students["Maths"].mode()
```

EXCERCISE ONE 
1. Find the mean science mark and explain your result 
2. FIND THE MEDIAN OF THE SCIENCE MARKS AND LIST THE STUDENTS ATTACHED TO THAT MARK


```python
students["Science"].mean()
```

The mean for the science marks is 78.33

```python
students["Science"].median()
```

The median is the number in the middle of a given dataset

EXCERSISE 2

```python
students["Science"].var()
```

The spread of the science marks using variance is 217.466
