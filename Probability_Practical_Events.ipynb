A python set beahves like a mathematical set, so it is a natural way to write a sample space.


```python
sample_space={1,2,3,4,5,6}
E={2,4,6}
A={x for x in sample_space if x>3}
B={x for x in sample_space if x<7}

print(f"Event E: {E}")
print (f"Event A: {A}")
print (f"Event B: {B}")
```

We can use the fractions packet to avoid decimals 

```python
from fractions import Fraction
```

```python
def prob(event, sample_space):
    event = set (event) & set(sample_space)
    return Fraction(len(event),len(sample_space))
```

```python
## the probability of getting an even number.
print(f"the probability of getting an even number is {prob(E,sample_space)}")
```

```python
print(prob(A,sample_space))
```

3. Inclass Research: Several coins: Building sample spaces with itertools.product


Listing outcomes by hand gets tiring quickly. itertools.product builds every combination for us 

What is the sample space for:
1. 3 coins ?
2. 4 coins ?
3. 2 dices ?


How many outcomes are there in each case ?

```python
from itertools import product
from fractions import Fraction
```

```python
coins3 = list(product('HT',repeat=3))
print(len(coins3))
```

```python
coins4 = list(product('HT',repeat=4))
print(len(coins4))
```

```python
dice2 = list (product(range(1,7),repeat=2))
print(len(dice2))
```

Excercise 3.1, part 2, bonus probabilities

```python
from itertools import product

coins3 = list(product("HT", repeat=3))
coins4 = list(product("HT", repeat=4))
dice2 = list(product(range(1, 7), repeat=2))

count = 0
for outcome in coins3:
    if outcome.count("H") == 2:
        count += 1
print("3 coins, exactly 2 heads:", count / len(coins3))

count = 0
for outcome in coins4:
    if outcome.count("H") == 2:
        count += 1
print("4 coins, exactly 2 heads:", count / len(coins4))

count = 0
for outcome in coins4:
    if outcome.count("H") >= 3:
        count += 1
print("4 coins, at least 3 heads:", count / len(coins4))

count = 0
for d1, d2 in dice2:
    if d1 + d2 == 7:
        count += 1
print("2 dice, sum is 7:", count / len(dice2))

count = 0
for d1, d2 in dice2:
    if d1 == d2:
        count += 1
print("2 dice, doubles:", count / len(dice2))

count = 0
for d1, d2 in dice2:
    if d1 + d2 >= 10:
        count += 1
print("2 dice, sum >= 10:", count / len(dice2))  
```
