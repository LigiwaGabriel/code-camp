MATPLOTLIB 

```python
import matplotlib.pyplot as plt
```

```python
x = [1, 2, 3, 4,]
y = [10,20,15,25]
plt.plot(x, y)
plt.title("Simple Line Graph")
plt.xlabel("X values")
plt.ylabel("Y values")
plt.show()
```

SEABORN

```python
import seaborn as sns
```

```python
data = [20, 25, 30, 35, 40]

sns.histplot(data=data)
plt.title("Distribution of Data")
plt.xlabel("Value")
plt.ylabel("Frequency")
plt.show()
```

PLOTLY

```python
import plotly.express as px
```

```python
student_data = {
    "Student": ["A", "B", "C", "D"],
    "Marks": [70, 85, 60, 90]
}

fig = px.bar(
    student_data,
    x="Student",
    y="Marks",
    title="Student Marks"
)

fig.show(renderer="browser")
```
