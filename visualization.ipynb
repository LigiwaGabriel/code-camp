PLOTLY

```python
import plotly.express as px
```

```python
student_data = {
    "Student": ["A", "B", "C", "D"],
    "Marks": [70, 85, 60, 90]
}

fig = px.bar(
    student_data,
    x="Student",
    y="Marks",
    title="Student Marks"
)

fig.show(renderer="browser")
```

```python
import numpy as np 
```

```python
import pandas as pd

sales = pd.DataFrame({
    "category": ["Electronics", "Clothing", "Groceries", "Books", "Toys"],
    "revenue": [45000, 28000, 32000, 9000, 14000]
})
```

```python
np.random.seed(42)
hours = np.random.uniform(1, 10, 50)
scores = 40 + 5 * hours + np.random.normal(0, 6, 50)
students = pd.DataFrame({"hours_studied": hours, "exam_score": scores})
```

```python
import matplotlib.pyplot as plt
plt.pie(
    sales["revenue"],
    labels=sales["category"],
    autopct="%1.1f%%",
    startangle=90
)
plt.title("Revenue by Category")
plt.axis("equal")
plt.show()
```
