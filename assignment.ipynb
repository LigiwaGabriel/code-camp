```python
import pandas as pd 
```

```python
students=pd.read_csv("StudentsPerformance.csv")
students
```

QN.1
How many female students scored 80 and above in mathematics?

```python
len(students[(students["gender"] == "female") & (students["math score"] >= 80)])
```

QN.2 What is the mean reading score of students who completed the test preparation course

```python
students[students["test preparation course"]=="completed"]["reading score"].mean()
```

QN.3 WHAT IS THE RANGE OF THE WRITING SCORE 

```python
students["writing score"].max()-students["writing score"].min()
```

QN.4 Find the standard deviation of the male students and their reading score

```python
students[students["gender"]=="male"]["reading score"].std()
```

QN.5 Compute the total score that is to say math, reading, writing

```python
students["total score"]=(
    students["math score"]+
    students["reading score"]+
    students["writing score"]
)
students.head()
```

a) Find the average of the total score

```python
students["total score"].mean()
```

b) the 90th percentile of the total score 

```python
students["total score"].quantile(0.90)
```

c) find the inter quartile range 

```python
students["total score"].quantile(0.75)-students["total score"].quantile(0.25)
```
